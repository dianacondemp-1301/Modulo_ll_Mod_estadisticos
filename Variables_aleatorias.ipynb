# Variables aleatorias

Sea $(\Omega, \mathcal{F}, P)$ un **espacio de probabilidad**.

Una variable aleatoria real se define como:

$$
X:\Omega \rightarrow \mathbb{R}
$$

Las variables aleatorias (v.a.) se clasifican en:

- **Discretas:** toman valores en algún conjunto discreto.
- **Continuas:** toman valores en algún intervalo de $\mathbb{R}$.
- **Mixtas:** son discretas y continuas.

## 1. Variables aleatorias discretas

Una variable aleatoria $X$ asigna a cada resultado $\omega \in \Omega$ un valor real:

$$
X(\omega_i)=x_i
$$
### Función de probabilidad

La función $p_X$ asigna masa de probabilidad a los valores que puede tomar $X$:

$$
p_X:\mathbb{R}\rightarrow[0,1]
$$

Se define como:

$$
p_X(k)=
\begin{cases}
P(X=k), & \text{si } k\in R_X \\[6pt]
0, & \text{en otro caso}
\end{cases}
$$

donde $R_X$ representa el conjunto de valores que puede tomar la variable aleatoria $X$.
### Propiedades

La función de probabilidad cumple:

**1. No negatividad y acotación**

$$
0\leq p_X(k)\leq 1
$$

**2. Normalización**

$$
\sum_{k\in R_X}p_X(k)=1
$$

## 2. Función de distribución acumulada

Existe una función que acumula probabilidades, llamada **función de distribución**.

Se define como:

$$
F_X(x):=P(X\leq x)
$$

Para una variable aleatoria discreta:

$$
\boxed{
F_X(x)=P(X\leq x)=\sum_{u\leq x}p_X(u)
}
$$


# Variables aleatorias absolutamente continuas (continuas)

Decimos que $X$ es una variable aleatoria continua si existe $f_X$ no negativa tal que:

$$
F_X(x)=\int_{-\infty}^{x}f_X(t)\,dt
$$

donde $f_X$ es la función de densidad:

$$
f_X:\mathbb{R}\rightarrow\mathbb{R}_{+}
$$

### Propiedades de la función de densidad

1. $f_X(x)\geq 0$

2. $\displaystyle\int_{\mathbb{R}}f_X(x)\,dx=1$



### Teorema Fundamental del Cálculo

Para una variable aleatoria continua:

$$
\frac{dF_X(x)}{dx}=f_X(x)
$$

## Esperanza matemática o valor esperado

Me interesa resumir la información de la variable aleatoria $X$ mediante un valor representativo (**esperanza matemática o valor esperado**).

Sea $X$ una variable aleatoria real, definimos la esperanza de $X$ como sigue:

$$
E[X]=
\begin{cases}
\displaystyle\sum_{k\in R_X}kP(X=k),
& X\text{ es discreta}
\\[12pt]
\displaystyle\int_{\mathbb{R}}x f_X(x)\,dx,
& X\text{ es continua}
\end{cases}
$$

La cual existe si:

**Para variables aleatorias discretas:**

$$
\sum_k |k|P(X=k)<\infty
$$

**Para variables aleatorias continuas:**

$$
\int_{\mathbb{R}}|x|f_X(x)\,dx<\infty
$$

**Nota:** La variable aleatoria de Cauchy no tiene esperanza definida.

La esperanza $E[X]$ es un número real (puede ser negativa).  

## Ejemplo: Lanzamiento de un dado

Sea $X$ el resultado del lanzamiento de un dado.

La función de probabilidad está dada por:

$$
p_X(k)=
\begin{cases}
\frac{1}{6}, & \text{si } k\in\{1,\ldots,6\}\\[6pt]
0, & \text{en otro caso}
\end{cases}
$$

La esperanza matemática es:

$$
E[X]=\sum_{k=1}^{6}kP(X=k)
$$

$$
=1\left(\frac{1}{6}\right)+2\left(\frac{1}{6}\right)+\cdots+6\left(\frac{1}{6}\right)
$$

$$
=\frac{1}{6}\sum_{i=1}^{6}i
=\frac{21}{6}=3.5
$$

La esperanza matemática representa el **centro de masa** de la distribución.

$$
\boxed{E[X]=3.5}
$$

### Propiedades de la esperanza matemática

- $E[a]=a$

- $E[X]\geq 0$, si $X\geq 0$

- $E[aX+b]=aE[X]+b$

## Varianza

Si ahora queremos saber qué tan dispersos están nuestros datos, utilizamos la **varianza**.

La varianza se define como:

$$
\boxed{
\operatorname{Var}(X):=E\left[(X-E[X])^2\right]
}
$$

La varianza mide la distancia que hay de los valores con respecto a su media.

Además:

$$
\operatorname{Var}(X)\geq 0
$$

### 1. Cálculo mediante transformación de variable

Definimos:

$$
Y=g(X)=(X-E[X])^2
$$

Calculamos la densidad o función de probabilidad de $Y$ por medio del **teorema de cambio de variable**:

$$
f_Y \quad \text{o} \quad p_Y
$$

Y después aplicamos la definición de esperanza matemática.

### 2. Ley del estadístico inconsciente

Podemos calcular $E[Y]$, con $Y=g(X)$, de la siguiente manera:

$$
\boxed{
E[Y]=E[g(X)]=
\begin{cases}
\displaystyle\sum_x g(x)P(X=x),
& X\text{ es discreta}
\\[12pt]
\displaystyle\int_{\mathbb{R}}g(x)f_X(x)\,dx,
& X\text{ es continua}
\end{cases}
}
$$

## Otra forma de calcular la varianza

La varianza también se puede calcular de la siguiente manera:

$$
\boxed{
\operatorname{Var}(X)=E[X^2]-(E[X])^2
}
$$

### Momentos de una variable aleatoria

**Segundo momento de $X$:**

$$
E[X^2]
$$

**Tercer momento de $X$:**

$$
E[X^3]
$$

$\vdots$

**Momento de orden $n$:**

$$
E[X^n]
$$

## Variable aleatoria uniforme discreta

Una variable aleatoria $X$ tiene distribución uniforme discreta en el conjunto $\{x_{1},\dots,x_{n}\}$ si su función de masa de probabilidades está dada por
$$f_{X}(k)=\mathbb{P}(X=k)=\begin{cases}
\frac{1}{n} &\quad \text{si } k\in\{x_{1},\dots,x_{n}\} \\
0 &\quad \text{en otro caso}
\end{cases}$$

**Notación:** $X\sim Unif(x_{1},\dots,x_{n}).$

Se llama uniforme porque cada uno de los posibles valores de $X$ tienen la misma probabilidad.

Si $X\sim Unif(x_{1},\dots,x_{n})$, en este caso, se tiene:
	$$\mathbb{E}[X] = \sum_{k}kf_{X}(k)=\frac{x_n + x_1}{2}.$$
	y
	$$\text{Var}(X) = \mathbb{E}[(X-\mathbb{E}[X])^{2}] =\sum_{k}(k-\mathbb{E}[X])^{2}P_{X}(k)=\frac{(x_n - x_1+1)^2-1}{12}$$
