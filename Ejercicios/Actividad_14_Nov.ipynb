$\color{red}{\text{Ejercicio 1.}}$ Verifica que, en efecto, la esperanza y varianza de una variable uniforme está dada como antes.

Sea $X$ una variable aleatoria uniforme discreta:

$$
X\sim Unif(x_1,\ldots,x_n)
$$

donde $x_1,\ldots,x_n$ son $n$ enteros consecutivos, con:

$$
x_n=x_1+n-1
$$

Su función de masa de probabilidad es:

$$
p_X(k)=P(X=k)=
\begin{cases}
\frac{1}{n}, & k\in\{x_1,\ldots,x_n\}\\[6pt]
0, & \text{en otro caso}
\end{cases}
$$

### 1. Demostración de la esperanza matemática

Por definición:

$$
E[X]=\sum_{k=x_1}^{x_n}k\,p_X(k)
$$

Sustituyendo $p_X(k)=\frac{1}{n}$:

$$
\begin{aligned}
E[X]
&=\sum_{k=x_1}^{x_n}k\left(\frac{1}{n}\right)\\[6pt]
&=\frac{1}{n}\sum_{k=x_1}^{x_n}k\\[6pt]
&=\frac{1}{n}\left(x_1+x_2+\cdots+x_n\right)
\end{aligned}
$$

Como los valores son enteros consecutivos, la suma de una progresión aritmética es:

$$
\sum_{k=x_1}^{x_n}k=\frac{n(x_1+x_n)}{2}
$$

Por lo tanto:

$$
\begin{aligned}
E[X]
&=\frac{1}{n}\left(\frac{n(x_1+x_n)}{2}\right)\\[6pt]
&=\frac{x_1+x_n}{2}
\end{aligned}
$$

Así, queda demostrado que:

$$
\boxed{E[X]=\frac{x_1+x_n}{2}}
$$

---

### 2. Demostración de la varianza

Por definición:

$$
Var(X)=E[(X-E[X])^2]
$$

También sabemos que:

$$
Var(X)=E[X^2]-(E[X])^2
$$

Primero calculamos el segundo momento:

$$
E[X^2]=\sum_{k=x_1}^{x_n}k^2p_X(k)
$$

Sustituyendo la función de probabilidad:

$$
\begin{aligned}
E[X^2]
&=\sum_{k=x_1}^{x_n}k^2\left(\frac{1}{n}\right)\\[6pt]
&=\frac{1}{n}\sum_{k=x_1}^{x_n}k^2
\end{aligned}
$$

Como $k=x_1+j-1$, con $j=1,\ldots,n$:

$$
\begin{aligned}
E[X^2]
&=\frac{1}{n}\sum_{j=1}^{n}(x_1+j-1)^2\\[6pt]
&=\frac{1}{n}\sum_{j=1}^{n}
\left[(x_1-1)^2+2(x_1-1)j+j^2\right]\\[6pt]
&=(x_1-1)^2
+\frac{2(x_1-1)}{n}\sum_{j=1}^{n}j
+\frac{1}{n}\sum_{j=1}^{n}j^2
\end{aligned}
$$

Utilizamos las fórmulas:

$$
\sum_{j=1}^{n}j=\frac{n(n+1)}{2}
$$

$$
\sum_{j=1}^{n}j^2=\frac{n(n+1)(2n+1)}{6}
$$

Sustituyendo:

$$
\begin{aligned}
E[X^2]
&=(x_1-1)^2
+\frac{2(x_1-1)}{n}\frac{n(n+1)}{2}
+\frac{1}{n}\frac{n(n+1)(2n+1)}{6}\\[6pt]
&=(x_1-1)^2
+(x_1-1)(n+1)
+\frac{(n+1)(2n+1)}{6}
\end{aligned}
$$

Por otra parte, como $x_n=x_1+n-1$:

$$
E[X]=\frac{x_1+x_n}{2}
=x_1+\frac{n-1}{2}
$$

Aplicamos la fórmula de la varianza:

$$
\begin{aligned}
Var(X)
&=E[X^2]-(E[X])^2\\[6pt]
&=(x_1-1)^2+(x_1-1)(n+1)
+\frac{(n+1)(2n+1)}{6}
-\left(x_1+\frac{n-1}{2}\right)^2\\[6pt]
&=\frac{n^2-1}{12}
\end{aligned}
$$

Finalmente, dado que:

$$
n=x_n-x_1+1
$$

Obtenemos:

$$
\begin{aligned}
Var(X)
&=\frac{n^2-1}{12}\\[6pt]
&=\frac{(x_n-x_1+1)^2-1}{12}
\end{aligned}
$$

Por lo tanto, queda demostrado que:

$$
\boxed{
Var(X)=\frac{(x_n-x_1+1)^2-1}{12}
}
$$

---

### Conclusión

Para una variable aleatoria uniforme discreta sobre $n$ enteros consecutivos:

**Esperanza matemática:**

$$
\boxed{E[X]=\frac{x_1+x_n}{2}}
$$

**Varianza:**

$$
\boxed{
Var(X)=\frac{(x_n-x_1+1)^2-1}{12}
}
$$

$\color{red}{\text{Ejercicio 2.}}$ Utilizar el atributo `.cdf` para verificar los resultados de arriba. Además gráficar la función de densidad de una Poisson

```python
# Parámetro de la distribución Poisson
lam = 2

# Probabilidad de que ocurran más de 2 accidentes
F2 = sps.poisson.cdf(2, mu=lam)  # P(X <= 2)
P1 = 1 - F2                     # P(X > 2)

# Probabilidad de que ocurra por lo menos 1 accidente
F0 = sps.poisson.cdf(0, mu=lam)  # P(X = 0)
P2 = 1 - F0                     # P(X >= 1)

# Probabilidad condicional P(X > 2 | X >= 1)
Proba = P1 / P2

print("P(X > 2) =", P1)
print("P(X > 2 | X >= 1) =", Proba)
```

```python
# Graficamos la Poisson

# Parámetros
lam = 2

# Generamos números aleatorios con distribución Poisson
poisson_numeros = sps.poisson.rvs(mu=lam, size=1000)

# Creamos un histograma
plt.figure(figsize=(8, 4))
plt.hist(
    poisson_numeros,
    density=True, # Normaliza para representar probabilidades
    bins=np.arange(poisson_numeros.max() + 2) - 0.5,
    color="#8BE1E9",
    edgecolor="k"
)

plt.xlabel("Valores de X")
plt.ylabel("Probabilidad")
plt.title("Distribución de Poisson")
plt.xticks(np.arange(poisson_numeros.max() + 1))
plt.show()
```
